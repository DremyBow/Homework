## Задание 7. Гистограмма

```python
plt.figure(figsize=(9, 5))
plt.hist(df["popularity"], bins=10, edgecolor="black")
plt.title("Распределение значений popularity")
plt.xlabel("Популярность")
plt.ylabel("Количество треков")
plt.xlim(0, 92)
plt.grid(axis="y", alpha=0.25)
plt.show()
```

**Комментарий по графику:** основная часть значений популярности находится примерно в диапазоне **0–46**, особенно в нижней и средней части шкалы. Значения в верхней части диапазона, примерно **74–92**, встречаются значительно реже остальных.


## Контрольные вопросы

**1. Что такое DataFrame?**  
DataFrame — это двумерная структура данных Pandas, состоящая из строк и столбцов.

**2. Что такое признак (Feature)?**  
Признак — это характеристика объекта, представленная отдельным столбцом в наборе данных.

**3. Чем среднее арифметическое отличается от медианы?**  
Среднее арифметическое рассчитывается как сумма всех значений, делённая на их количество. Медиана — это центральное значение упорядоченного ряда данных. Медиана меньше зависит от очень больших или очень маленьких значений.

**4. Что показывает гистограмма?**  
Гистограмма показывает распределение числовых значений по интервалам и количество наблюдений в каждом интервале.

**5. Для чего нужна фильтрация данных?**  
Фильтрация позволяет выбрать только те строки, которые соответствуют заданному условию, например треки, выпущенные после 2024 года.
