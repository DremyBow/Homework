# Практическая работа №3


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Если файл загружен в Colab с таким именем:
df = pd.read_csv('vgsales.csv')
df.head(10)
```

### Задание 2. Общая информация о данных

```python
print('Количество строк:', df.shape[0])
print('Количество столбцов:', df.shape[1])
print('\nНазвания столбцов:')
print(df.columns.tolist())
print('\nТипы данных:')
print(df.dtypes)
print('\nОсновные статистические характеристики:')
display(df.describe())
```

**Ответ:** числовые столбцы — `Rank`, `Year`, `NA_Sales`, `EU_Sales`, `JP_Sales`, `Other_Sales`, `Global_Sales`. Текстовые — `Name`, `Platform`, `Genre`, `Publisher`. Столбцы с продажами — `NA_Sales`, `EU_Sales`, `JP_Sales`, `Other_Sales`, `Global_Sales`.

### Задание 3. Поиск пропущенных значений

```python
missing = df.isnull().sum()
display(missing)
print('Строки, в которых отсутствует Year:')
display(df[df['Year'].isnull()])
```

**Ответ:** пропуски есть в `Year` — 271 значений, и в `Publisher` — 58 значений. Больше всего пропусков находится в `Year`.

### Задание 4. Очистка данных

```python
df_clean = df.dropna(subset=['Year'])
print('Размер исходного df:', df.shape)
print('Размер df_clean:', df_clean.shape)
print('\nПропуски после удаления строк без Year:')
display(df_clean.isnull().sum())
```

**Ответ:** в исходном датасете 16598 строк. После удаления строк без `Year` осталось 16327 строк. Пропуски в `Year` мешают анализу игр по годам, потому что такие записи нельзя корректно отнести к конкретному году выпуска. Исходный `df` не изменяется; дальнейший анализ выполняется на `df_clean`.

### Задание 5. Уникальные значения категориальных признаков

```python
print('Жанры:')
print(df_clean['Genre'].unique())
print('Количество жанров:', df_clean['Genre'].nunique())
print('\nКоличество платформ:', df_clean['Platform'].nunique())
print('Количество издателей:', df_clean['Publisher'].nunique())
```

**Ответ:** представлено 12 жанров, 31 платформ и 576 издателей. Больше всего уникальных значений имеет признак `Publisher`.

### Задание 6. Количество игр каждого жанра

```python
genre_count = df_clean.groupby('Genre')['Name'].count().sort_values(ascending=False)
display(genre_count)
```

### Задание 7. Мировые продажи по жанрам

```python
genre_sales = df_clean.groupby('Genre')['Global_Sales'].sum().sort_values(ascending=False)
display(genre_sales)
```

### Задание 8. 10 самых продаваемых игр

```python
top_games = df_clean.nlargest(10, 'Global_Sales')[['Rank', 'Name', 'Platform', 'Genre', 'Global_Sales']]
display(top_games)
```

### Задание 9. Анализ платформ

```python
platform_count = df_clean.groupby('Platform')['Name'].count().sort_values(ascending=False)
print('10 платформ с наибольшим количеством игр:')
display(platform_count.head(10))

platform_sales = df_clean.groupby('Platform')['Global_Sales'].sum().sort_values(ascending=False)
print('Платформы с наибольшими суммарными продажами:')
display(platform_sales.head(10))
```

### Задание 10. Визуализация продаж по жанрам

```python
plt.figure(figsize=(10, 6))
genre_sales.sort_values(ascending=True).plot(kind='barh')
plt.title('Мировые продажи по жанрам')
plt.xlabel('Global Sales, млн')
plt.ylabel('Жанр')
plt.tight_layout()
plt.show()
```

**Ответ:** наибольшие суммарные продажи у жанра `Action` (1722.88 млн), затем у `Sports` (1309.24 млн). Наименьшие — у `Strategy` (173.43 млн) и `Adventure` (234.80 млн). Значения заметно различаются: суммарные продажи Action примерно в 10 раз выше Strategy.

### Задание 11. Корреляция NA_Sales и Global_Sales

```python
corr = df_clean[['NA_Sales', 'Global_Sales']].corr()
display(corr)
print('Коэффициент корреляции:', corr.loc['NA_Sales', 'Global_Sales'])
```

**Ответ:** коэффициент корреляции равен примерно **0.941**. Он положительный и близок к 1, поэтому по данным наблюдается выраженная положительная линейная взаимосвязь между продажами в Северной Америке и мировыми продажами. При этом корреляция сама по себе не доказывает причинно-следственную связь.

### Задание 12. Диаграмма рассеяния

```python
plt.figure(figsize=(8, 5))
plt.scatter(df_clean['NA_Sales'], df_clean['Global_Sales'], alpha=0.5)
plt.title('Связь продаж в Северной Америке и мировых продаж')
plt.xlabel('Продажи в Северной Америке, млн')
plt.ylabel('Мировые продажи, млн')
plt.tight_layout()
plt.show()
```

**Ответ:** одна точка соответствует одной игре из датасета. На графике наблюдается восходящая тенденция: при увеличении продаж в Северной Америке обычно увеличиваются и мировые продажи. Это соответствует результату корреляционного анализа, где коэффициент близок к 1.

### Задание 13. Корреляция между региональными продажами

```python
region_corr = df_clean[['NA_Sales', 'EU_Sales', 'JP_Sales', 'Other_Sales']].corr()
display(region_corr)
```

**Ответ:** самая сильная взаимосвязь между региональными продажами наблюдается между **NA_Sales и EU_Sales**: коэффициент корреляции ≈ **0.769**.
